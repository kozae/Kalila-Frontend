```python
import unicodedata
import pyperclip


def normalize_text(text):   
    return unicodedata.normalize('NFC', text)


def copy_to_clipboard(string_list, index):
    if index < len(string_list):
        pyperclip.copy(string_list[index])
    else:
        print("Index out of range.")
```

```python
text = """
إنّ الرجل
وإن كان شجاعًا سخيّا ثمّ لم يكن حليمًا عاقلًا وكان مشاوره
غير لبيب فإنّه يبهظه الأمر اليسير لضعف رأيه وجهالته
وخطأ رأي نصحائه وإن أصاب ظفرًا أو لقى رشدًا لقدر ساقه إليه
صارت عاقبة أمره إلى الندامة
"""
```

```python
from IPython.display import Markdown
text = [x for x in normalize_text(text).splitlines() if len(x) > 0]
text
```
